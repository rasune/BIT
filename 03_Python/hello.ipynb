```python
skaicius = int(input("Įveskite skaičių: "))
if skaicius == 0:
    print("Skaičius yra nulis")
elif skaicius % 2 == 0:
    print("Skaičius yra lyginis")
else:
    print("Skaičius yra nelyginis")
```
