1.Creat a list of numbers from 1 to 10. Use FOR cycle and print just even numbers. Use continue

```python
numbers = list(range(1,11))

for number in numbers:
    if number %2 != 0:
        continue    #liepia ciklui praleisti likusi sios iteracijos koda ir eik prie kito elemento
    print(number)
```

2.You have a list: ['apple', 'banana', 'cherry', 'date']. Use a `for` loop and an `if` to print out the fruits name length is greater than 5.

```python
fruits =['apple', 'banana', 'cherry', 'date']

for fruit in fruits:
   if len(fruit) > 5:
    print(fruit)
```

3. Use a `while` loop to count the sum of the numbers from 1 to 20. If the sum exceeds 100, terminate the loop with `break`.

```python
number = 1
suma = 0

while number <= 20:
    suma += number
    if suma > 100:
     break
number += 1
print("Suma:",suma)
```

4. You have a dictionary with student grades. Print only those students whose grade is greater than or equal to 8. Use `if` and `for`.

```python
students = {
    "Anna": 10,
    "John": 5,
    "Paul": 9,
    "Mia": 7
}

for stu_name, stu_mark in students.items():
    print(stu_name)
    print(stu_mark)
```

7. Naudodami `for` ciklą padvigubinkite kiekvieną sąrašo reikšmę: [2, 4, 6]. Rezultatas turi būti
naujas sąrašas.
Using a 'for' loop, double each value in the list [2, 4, 6]. The result must be a new list

```python
numbers = [2, 4, 6]
new = []

for x in numbers:
    new.append(x * 2)
    
print(new)
```
